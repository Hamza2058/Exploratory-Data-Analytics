```
# Essential Python Libraries are imported for cleaning, visualisations etc.
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

# Essential Data Cleaning begins:

df = pd.read_csv("shopping_trends.csv") # File is imported into Colab and now assigned into a dataframe
# Data Inspection Process
df.describe()
df.info()
df.shape

# Potential Data Cleaning
df.duplicated()
df.isna().sum()
df.columns
# No abnormalities or duplicates/Null values were found.


# Exploratory Data Analysis begins:
df.groupby("Review Rating")["Purchase Amount (USD)"].sum().head(10).plot(kind = "bar", color = "Red") # Graph showing how Purchase Amount increases with better reviews
plt.title("Purchase Amount in relation to Review Rating") # Graph title
plt.xlabel("Purchase Amount") # X label
plt.ylabel("Review Rating") # Y label
plt.show()

df["Shipping Type"].value_counts().plot(kind = "bar", color = "skyblue") # bar chart showing most frequent shipping types
plt.xlabel("Shipping Type") # X label
plt.ylabel("Frequency") # Y label
plt.title("Most used Shipping Types") # Graph title
plt.show()

df["Gender"].value_counts().plot(kind = "bar", color = "Pink") # Bar chart showing Gender Frequency
plt.xlabel("Gender") # X label
plt.ylabel("Frequency") # Y label
plt.title("Gender Distribution") # Graph title
plt.show()

df["Payment Method"].value_counts().plot(kind = "bar", color= "Green",)# Bar chart showing most frequent payment methods
plt.xlabel("Payment Method") # X label
plt.ylabel("Frequency") # Y label
plt.title("Payment Method distribution") # Graph title
plt.show()

df["Category"].value_counts().plot(kind = "line", color = "Blue", marker= ".", markersize= "15" )# Line chart showing most frequent Categories for orders
plt.ylabel("Frequency") # Y label
plt.xlabel("Item Category") # X label
plt.title("Most Frequent Category Ordered") # Graph title
plt.show()

df["Subscription Status"].value_counts().plot(kind = "barh", color="Orange")# Bar chart showing if order was done with or without Subscription
plt.title("How many orders were placed with a Subscription") # Graph title
plt.xlabel("Frequency") # X label
plt.ylabel("Subscription Status") # Y label
plt.show()

sns.boxplot(x = "Category", y = "Purchase Amount (USD)", data = df)# Boxplot showing mean purchase amount etc for each Category
plt.title("Mean Purchase Amount based on each Category") # Graph title
plt.show()

sns.displot(df["Age"], bins = 40, color = "cyan") # Distribution plot showing Age
plt.title("Age distribution of orders") # Graph title
plt.show()
```

Purchase Amount vs Review Rating

The bar chart shows how total purchase amounts vary across review ratings. Higher review ratings generally correspond to greater total purchase values, suggesting that positive customer feedback is associated with increased spending and repeat purchases.

Shipping Type Usage

The shipping type distribution indicates that certain shipping methods are used far more frequently than others. This suggests customer preference for specific delivery options, likely influenced by cost, speed, or convenience.

Gender Distribution

The gender distribution chart shows an imbalance in order frequency between genders. This indicates that one gender group places more orders, highlighting a potential target demographic for marketing strategies.

Payment Method Distribution

The bar chart reveals that a small number of payment methods dominate customer transactions. This suggests that customers prefer familiar and convenient payment options, while alternative methods see lower adoption.

Category Frequency

The line plot of item categories shows that certain product categories are ordered more frequently than others. This highlights key revenue-driving categories and indicates areas where inventory and promotions can be prioritized.

Subscription Status

The horizontal bar chart shows that non-subscription purchases exceed subscription-based orders. This suggests that while subscriptions exist, most customers still prefer one-time purchases, indicating growth potential for subscription models.

Purchase Amount by Category

The boxplot illustrates variation in purchase amounts across product categories. Some categories show higher median spending and greater spread, indicating higher-value purchases, while others are more consistent and lower priced.

Age Distribution of Customers

The age distribution histogram shows that orders are concentrated within specific age ranges. This suggests that purchasing activity is driven primarily by a particular age group, which can inform targeted marketing and product positioning.

```
# Essential Python Libraries are imported for cleaning, visualisations etc.
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```
# Essential Data Inspection Begins

df = pd.read_csv("dataset_olympics.csv") # Dataset is loaded into Collab

df.describe() # Inspecting Dataframe description
df.info() # Inspecting Dataframe information
df.shape # Inspecting dataframe structure
df.head() # Inspecting first 5 rows
```

```
# Essential Data Cleaning begins

df.duplicated().sum() # 383 duplicates are detected
df = df.drop_duplicates() # All duplicates are removed.
df.isna().sum() # Detecting all NaN values in database
# Many NaN values are detected but are unique to each Athlete hence not filled with mean or "Missing".
# No critical values are missing/Nan
```

```
# Exploratory Data Analytics begins

df["Sex"].value_counts().plot(kind = "bar", color = "Red")
plt.ylabel("No. of")
plt.title("Gender Distribution of Athletes")
plt.show()

df.groupby("Sex")["Medal"].value_counts().plot(kind = "bar", color = "Gold")
plt.title("Medals per Gender")
plt.ylabel("Number of Medals")
plt.show()

sns.kdeplot(df["Age"], color = "Green",  marker = ".")
plt.title("Distribution of Athlete Age")
plt.show()

plt.scatter(x = df["Age"].head(700), y = df["Height"].head(700), color = "Purple")
plt.ylabel("Height")
plt.xlabel("Age")
plt.title("Distribution of Athelete Height according to their Age")
plt.show()

df["Sport"].value_counts().plot(kind = "line", color = "Green", linewidth = "3",)
plt.xlabel("Count")
plt.title("No. of Athletes per Sport")
plt.show()

df.groupby("Medal")["Team"].value_counts().head(4).plot(kind = "line", color = "Cyan")
plt.title("Medals per Team")
plt.ylabel("Number of Medals")
plt.show()

df["City"].value_counts().plot(kind = "line", color = "Orange")
plt.ylabel("Number of Athletes")
plt.title("City with the most Athletes")
plt.show()

df["NOC"].value_counts().plot(kind = "line", color = "Indigo")
plt.title("Country with the most Athletes")
plt.ylabel("No. of Athletes")
plt.show()
```



📊 Gender Distribution of Athletes

Male athletes significantly outnumber female athletes, indicating historical gender imbalance in Olympic participation.

Female participation has increased over time but remains lower overall.

🏅 Medals per Gender

Male athletes have won more medals overall, largely reflecting higher participation rates.

Medal distribution suggests competitive parity when adjusted for participation size.

📈 Distribution of Athlete Age

Athlete ages are concentrated in early to mid-20s, indicating peak physical performance years.

Very young and older athletes are comparatively rare.

📐 Athlete Height vs Age

No strong correlation between age and height is observed.

Height appears more sport-specific than age-dependent.

🏟 Number of Athletes per Sport

A small number of sports account for a large share of athletes.

Popular team sports dominate athlete representation compared to niche disciplines.

🏅 Medals per Team

Medal wins are concentrated among a few high-performing teams.

This reflects strong sports infrastructure and consistent elite performance.

🌍 City with the Most Athletes

Host or historically prominent Olympic cities show higher athlete representation.

Indicates geographical clustering of Olympic participation.

🌐 Country with the Most Athletes (NOC)

A few countries contribute the majority of athletes.

Larger populations and better funding correlate with higher participation.

🔍 Overall Insight

Olympic participation and success are unevenly distributed across gender, countries, and sports, driven by resources, infrastructure, and historical access.

**This analysis highlights participation inequality and performance concentration across demographics, geography, and sporting disciplines.**
